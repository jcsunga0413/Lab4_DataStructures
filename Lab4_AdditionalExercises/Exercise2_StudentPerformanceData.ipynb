```python

from collections import defaultdict
from statistics import mean
from pprint import pprint

LAST_NAME = "Sunga"
SEED_NUM = 5
FAVORITE_ARTIST = "The Neighbourhood"

SUBJECTS = [
    "PROGRAMMING", "MATHEMATICS", "ELECTRONICS",
    "PHYSICS", "ENGLISH"
]
STUDENTS = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

artist_values = [ord(char) for char in FAVORITE_ARTIST if char.isalpha()]
records = []

for i, student in enumerate(STUDENTS):
    for j, subject in enumerate(SUBJECTS):
        score = 50 + (
            SEED_NUM + artist_values[(i + j) % len(artist_values)]
        ) % 51
        records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

# Include one repeated student-subject record.
records.append(records[0].copy())

# Include one incomplete record.
records.append({
    "student": STUDENTS[-1],
    "subject": SUBJECTS[-1],
    "score": None
})

def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 75:
        return "PASSED"
    return "NEEDS IMPROVEMENT"

for record in records:
    record["category"] = performance_category(record["score"])

organized = defaultdict(dict)
for record in records:
    organized[record["student"]][record["subject"]] = record["score"]

seen = set()
repeated = []
for record in records:
    key = (record["student"], record["subject"])
    if key in seen:
        repeated.append(record)
    else:
        seen.add(key)

incomplete = [r for r in records if r["score"] is None]
valid_records = [r for r in records if r["score"] is not None]

print("EXERCISE 2: STUDENT PERFORMANCE DATA ANALYZER")
print("\nGENERATED PERFORMANCE RECORDS:")
pprint(records)

print("\nORGANIZED BY STUDENT AND SUBJECT:")
pprint(dict(organized))

print("\nREPEATED RECORDS:")
pprint(repeated)

print("\nINCOMPLETE RECORDS:")
pprint(incomplete)

print("\nPERFORMANCE CATEGORIES:")
for record in valid_records:
    print(record["student"], record["subject"],
          record["score"], record["category"])

print("\nSTUDENT SUMMARIES:")
for student in STUDENTS:
    scores = [
        r["score"] for r in valid_records
        if r["student"] == student
    ]
    if scores:
        print(f"{student}: average={mean(scores):.2f}, "
              f"highest={max(scores)}, lowest={min(scores)}")

print("\nSUBJECT SUMMARIES:")
for subject in SUBJECTS:
    scores = [
        r["score"] for r in valid_records
        if r["subject"] == subject
    ]
    if scores:
        print(f"{subject}: average={mean(scores):.2f}, "
              f"records={len(scores)}")

print("\nORDERED RESULTS:")
pprint(sorted(valid_records, key=lambda r: (
    r["student"], r["subject"]
)))

print("\nFINAL SUMMARY")
print("Total records:", len(records))
print("Valid scores:", len(valid_records))
print("Incomplete records:", len(incomplete))
print("Repeated student-subject records:", len(repeated))
```
