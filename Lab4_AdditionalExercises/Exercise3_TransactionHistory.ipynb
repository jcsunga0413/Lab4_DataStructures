```python

from collections import defaultdict
from pprint import pprint

LAST_NAME = "Sunga"
SEED_NUM = 5
FAVORITE_ARTIST = "The Neighbourhood"

CUSTOMERS = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
PRODUCTS = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]

# Fixed product prices
PRICES = {
    "NOTEBOOK": 35.00,
    "PEN": 15.00,
    "USB": 250.00,
    "HEADPHONES": 450.00,
    "CALCULATOR": 180.00
}

artist_values = [ord(char) for char in FAVORITE_ARTIST if char.isalpha()]
records = []

for i, customer in enumerate(CUSTOMERS):
    for j, product in enumerate(PRODUCTS):
        quantity = 1 + (
            SEED_NUM + artist_values[(i + j) % len(artist_values)]
        ) % 5

        records.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "price": PRICES[product],
            "total": round(quantity * PRICES[product], 2)
        })

# Include one repeated customer-product transaction.
records.append(records[0].copy())

customer_summary = defaultdict(float)
product_summary = defaultdict(float)

for record in records:
    customer_summary[record["customer"]] += record["total"]
    product_summary[record["product"]] += record["total"]

seen = set()
repeated = []

for record in records:
    key = (record["customer"], record["product"])
    if key in seen:
        repeated.append(record)
    else:
        seen.add(key)

distinct_products = sorted(set(r["product"] for r in records))
distinct_customers = sorted(set(r["customer"] for r in records))
ordered = sorted(
    records,
    key=lambda r: (r["customer"], r["product"])
)

print("EXERCISE 3: TRANSACTION HISTORY ANALYZER")
print("\nGENERATED TRANSACTION RECORDS:")
pprint(records)

print("\nORGANIZED TRANSACTIONS:")
pprint(ordered)

print("\nREPEATED TRANSACTIONS:")
pprint(repeated)

print("\nDISTINCT CUSTOMERS:", distinct_customers)
print("DISTINCT PRODUCTS:", distinct_products)

print("\nCUSTOMER SUMMARIES:")
for customer, total in sorted(customer_summary.items()):
    print(f"{customer}: total purchases = {total:.2f}")

print("\nPRODUCT SUMMARIES:")
for product, total in sorted(product_summary.items()):
    print(f"{product}: total sales = {total:.2f}")

print("\nORDERED RESULTS:")
pprint(ordered)

print("\nOVERALL TRANSACTION SUMMARY")
print("Total transaction records:", len(records))
print("Unique customer-product pairs:", len(seen))
print("Repeated extra records:", len(records) - len(seen))
print("Total sales:", f"{sum(r['total'] for r in records):.2f}")
print("Distinct customers:", len(distinct_customers))
print("Distinct products:", len(distinct_products))
```
