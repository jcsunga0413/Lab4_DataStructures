```python

from collections import defaultdict
from statistics import mean
from pprint import pprint

LAST_NAME = "Sunga"
SEED_NUM = 5
FAVORITE_ARTIST = "The Neighbourhood"

CATEGORIES = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
SOURCES = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

# Generate reproducible values from artist character positions.
artist_values = [ord(char) for char in FAVORITE_ARTIST if char.isalpha()]
records = []

for i, source in enumerate(SOURCES):
    for j, category in enumerate(CATEGORIES):
        value = SEED_NUM + artist_values[(i + j) % len(artist_values)] % 100
        records.append({
            "source": source,
            "category": category,
            "value": value
        })

# Add one repeated source/category record.
records.append(records[0].copy())

# Organize records by source and category.
organized = defaultdict(lambda: defaultdict(list))
for record in records:
    organized[record["source"]][record["category"]].append(record["value"])

repeated = [
    record for record in records
    if sum(r["source"] == record["source"]
           and r["category"] == record["category"]
           and r["value"] == record["value"] for r in records) > 1
]
distinct = sorted(set(
    (r["source"], r["category"], r["value"]) for r in records
))

category_summary = defaultdict(list)
for record in records:
    category_summary[record["category"]].append(record["value"])

print("EXERCISE 1: ENGINEERING TEST DATA ORGANIZER")
print("\nGENERATED TEST DATA:")
pprint(records)

print("\nORGANIZED DATA BY SOURCE AND CATEGORY:")
pprint(dict(organized))

print("\nREPEATED RECORDS:")
pprint(list({(r["source"], r["category"], r["value"])
             for r in repeated}))

print("\nDISTINCT RECORD COUNT:", len(distinct))

print("\nCATEGORY SUMMARIES:")
for category, values in sorted(category_summary.items()):
    print(f"{category}: count={len(values)}, "
          f"min={min(values)}, max={max(values)}, "
          f"average={mean(values):.2f}")

print("\nPROCESSED RESULTS (SORTED BY CATEGORY, SOURCE, VALUE):")
pprint(sorted(records, key=lambda r: (
    r["category"], r["source"], r["value"]
)))

print("\nFINAL SUMMARY")
print("Total records:", len(records))
print("Unique records:", len(distinct))
print("Repeated extra records:", len(records) - len(distinct))
print("Categories:", len(CATEGORIES))
print("Sources:", len(SOURCES))
```
